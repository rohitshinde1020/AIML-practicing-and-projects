```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]

sales = [10, 20, 30, 40, 50]
profit = [5, 12, 20, 28, 35]

plt.plot(x, sales, label="Sales")
plt.plot(x, profit, label="Profit")

plt.legend(loc="upper left",fontsize = 12)
plt.grid(color="grey",linestyle= ':',linewidth=1)
plt.show()
```

```python
import matplotlib.pyplot as plt
import pandas as pd

df = pd.read_csv("Heart_dataset - Heart_dataset.csv")

plt.bar(df["ChestPain"],df["Age"],color="c")
plt.xlabel("ChestPain Categories")
plt.ylabel("Age comparison")

plt.show()
```

```python
import matplotlib.pyplot as plt
import pandas as pd

df = pd.read_csv("Heart_dataset - Heart_dataset.csv")

plt.scatter(df["Age"], df["RestBP"],alpha = 0.7,label="AGE vs RESTBP")

plt.annotate(
    "Patient",
    xy=(63, 145),
    xytext=(50, 160),
    arrowprops=dict(arrowstyle="->")
)

plt.xlabel("Age")
plt.ylabel("Resting Blood Pressure")
plt.title("Age vs Resting Blood Pressure")
plt.legend()

plt.show()
```

```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

df = pd.read_csv("Heart_dataset - Heart_dataset.csv")

plt.hist(df["Age"], bins=10,width=3,  edgecolor="black")


plt.axvline(
    np.mean(df["Age"]),
    linestyle="--"
)
plt.xticks(
    [20, 30, 40, 50, 60, 70],
    ["20 yrs", "30 yrs", "40 yrs", "50 yrs", "60 yrs", "70 yrs"]
)
plt.xticks(rotation=45)

plt.xlabel("Age")
plt.ylabel("Number of patients")
plt.title("Distribution of Age")

plt.show()
```

```python
import matplotlib.pyplot as plt
import pandas as pd

df = pd.read_csv("Heart_dataset - Heart_dataset.csv")

df["ChestPain"].value_counts().plot.pie(
    autopct="%1.1f%%"
)

plt.title("Chest Pain Type Distribution")
plt.ylabel("")
plt.show()
```

```python
df["Sex"].value_counts().plot.pie(
    autopct="%1.1f%%"
)
plt.title("SEX RATIO")
plt.ylabel("")
plt.show()
```

```python
import matplotlib.pyplot as plt

counts = df["ChestPain"].value_counts()

plt.pie(
    counts,
    labels=counts.index,
    autopct="%1.1f%%",
    explode=[0.1, 0.1, 0, 0]
)

plt.title("EXPLODE PIE CHART")
plt.show()
```

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(2, 2)

ax[0, 0].plot([1, 2, 3], [1, 4, 9])
ax[0, 0].set_title("Plot 1")

ax[0, 1].plot([1, 2, 3], [1, 2, 3])
ax[0, 1].set_title("Plot 2")

ax[1, 0].bar(["A", "B", "C"], [10, 20, 15])
ax[1, 0].set_title("Plot 3")

ax[1, 1].scatter([1, 2, 3], [3, 1, 2])
ax[1, 1].set_title("Plot 4")

plt.tight_layout()
plt.show()
```
